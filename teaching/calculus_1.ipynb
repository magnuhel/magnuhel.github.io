## The bisection method / Midtpunktmetoden 

I will first share the result of the coding I did live in class, before I present some improvements. 

### Example 1 

Let $f(x)$ be the function  

$$
f(x)=\cos(x)-x
$$

Find $x$ such that $f(x)\approx 0$.

```python
import math

#import matplotlib.pyplot as plt
import numpy as np

#MIDTPUNKTMETODEN f(x) kontinuerlig på [a,b] f(a) og f(b) forskjellige fortegn 
#Bisection method, f(x) continuous on [a,b], f(a) og f(b) opposite signs  

def f(x): 
    return math.cos(x) - x
    #return math.exp(x)-math.sin(x)-10

a = 0
b = math.pi #3.1415926535 8979323846
#b=4

#t = 1e-7
t=1e-3

if ( f(a) > 0 and f(b) > 0 ) or ( f(a) < 0 and f(b) < 0 ):
    print("Kriteriene for skjæringssetningen er ikke oppfylt")
    exit()

while b - a > t :
    c = (b+a)/2 #midtpunkt

    if ( f(a) > 0 and f(c) < 0 ) or ( f(a) < 0 and f(c) > 0 ): #f(a) og f(c) forskjellig fortegn
        b = c

    else:
        a = c

    print(a,b)

    #print(a,b)
```

### Improvement 

Using functions improves the code above.  

```python
import math

def bisec(f,a,b,tol=1e-3,verbose=False): #b-a>0 , f continuous , f monotone => unique solution  

    #Safety checks 
    if f(a) == 0: return a
    if f(b) == 0: return b 
    if f(a)*f(b)>0: raise ValueError("f(a) and f(b) must have opposite signs")

    while b-a > tol: # bisects until tol 
        c=(a+b)/2 #midpoint 
        if f(c) == 0: return c 
        if f(a)*f(c)<0: b=c #f(a) and f(c) on opposite sides
        else: a=c 
        if verbose: print("a=",a,"b=",b)
    return (a+b)/2 #returning midpoint, "for free" haha 

def g(x): #Ex 1
    return math.cos(x) - x
def h(x): #Ex 2
    return math.exp(x)-math.sin(x)-10 

#Example 1: 
print(bisec(g,0,math.pi))

#Example 2: 
print(bisec(h,0,4))
```

## Newton's method 

### Example 1

Let $f(x)=\cos(x)-x$. Find $x$ such that $f(x)\approx 0$. 


```python
import math

def f(x):
    return math.cos(x)-x

def df(x):
    return -math.sin(x)-1

x = 0

while abs(f(x))>1e-8:
    x=x-f(x)/df(x)
    print("x=",x,"f(x)=",f(x))
```

Notice how quickly the method converges: only a few iterations! 

### Example 2

$f(x)=e^x-\sin(x)-10$

```python
import math

def f(x):
    return math.exp(x)-math.sin(x)-10

def df(x):
    return math.exp(x)-math.cos(x)

x = 4 #x_0 , Skjæringssetningen [a=0 , b=4]

while abs(f(x))>1e-8:
    x=x-f(x)/df(x)
    print("x=",x,"f(x)=",f(x))
```

### Example 3 

$f(x)=\ln(x)+x$

```python
import math

def f(x):
    return math.log(x)+x

def df(x):
    return 1/x+1

x = 1/math.e #x_0 , Skjæringssetningen [a=e^-1 , b=e]

while abs(f(x))>1e-8:
    x=x-f(x)/df(x)
    print("x=",x,"f(x)=",f(x))
```

### Example 4

We will not necessarily find the zero closest to our starting point $x_0$. This can happen if we choose an $x_0$ close to a critical point. Let's illustrate that with the following example $f(x)=\cos(x)$: 

```python
import math

def f(x):
    return math.cos(x)

def df(x): #derivert 
    return -math.sin(x)

x = 1e-5 #very close to x=0

while abs(f(x))>1e-8:
    x=x-f(x)/df(x)
    print("x=",x,"f(x)=",f(x))
```

### Improvement 



```python
import math 

def nm(f,df,x,tol=1e-8,verbose=False,max_it=50):
    if f(x)==0: return x
    if df(x)==0: raise ValueError("Newton step undefined")
    it=0

    while abs(f(x))>tol:
        if it>=max_it: raise RuntimeError(f"no convergence after {max_it} iterations")
        #
        x=x-f(x)/df(x) 
        #
        if df(x)==0: raise ValueError("Newton step undefined") 
        if verbose: print("x=",x,"f(x)=",f(x))
        it+=1

    return x 

#Example 1 
def f1(x):
    return math.cos(x)-x
def df1(x):
    return -math.sin(x)-1

#Example 2 
def f2(x):
    return math.exp(x)-math.sin(x)-10
def df2(x):
    return math.exp(x)-math.cos(x)

#Example 3
def f3(x):
    return math.log(x)+x
def df3(x):
    return 1/x+1

#Example 4 
def f4(x):
    return math.cos(x)
def df4(x): 
    return -math.sin(x)

#Example Maximal Iterations 
def f5(x):
    return x**3-2*x+2
def df5(x):
    return 3*x**2-2

print(nm(f1,df1,0))
print(nm(f2,df2,4))
print(nm(f3,df3,1/math.e))
print(nm(f4,df4,1e-5))

#Showcase for f5: it's cycling between 0 and 1 
#print(nm(f5,df5,0,max_it=10,verbose=True))
try:
    nm(f5, df5, 0, max_it=10, verbose=True)
except RuntimeError as e:
    print("Stopped:", e)

```
